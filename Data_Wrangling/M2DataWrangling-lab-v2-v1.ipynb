### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>


```python
## Write your code here
df.isnull().sum().idxmax() # or df.isna().sum().idxmax()
```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


```python
## Write your code here
df.describe()
df['ConvertedCompYearly'] = df['ConvertedCompYearly'].fillna(df['ConvertedCompYearly'].mean())
df['WorkExp'] = df['WorkExp'].fillna(df['WorkExp'].median())
#df['WorkExp'].isna().sum()
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
## Write your code here
df['RemoteWork'].value_counts()
df['RemoteWork'] = df['RemoteWork'].fillna(df['RemoteWork'].value_counts().index[0])
#df['RemoteWork'].isna().sum()
```

### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


```python
## Write your code here
# Mathematical Formula
df['ConvertedCompYearly_MinMax'] = (df['ConvertedCompYearly'] - df['ConvertedCompYearly'].min()) / (df['ConvertedCompYearly'].max() - df['ConvertedCompYearly'].min())
df
```

<h5>6.2 Log-transform the ConvertedCompYearly column to reduce skewness.</h5>


```python
## Write your code here
import numpy as np
df['ConvertedCompYearly'] = np.log(df['ConvertedCompYearly'] + 1)
#df['ConvertedCompYearly']
```

### 7. Feature Engineering


<h5>7.1 Create a new column `ExperienceLevel` based on the `YearsCodePro` column:</h5>


```python
## Write your code here
#df['YearsCodePro'].unique()
#df['YearsCodePro'] = (
#    df['YearsCodePro'].replace(
#        {"Less than 1 year": 0, "More than 50 years" : 51}
#    ).astype(float)
#)
#df['YearsCodePro'].unique()
conditions = [
    df['YearsCodePro'] < 2,
    (df['YearsCodePro'] >= 2) & (df['YearsCodePro'] < 5),
    (df['YearsCodePro'] >= 5) & (df['YearsCodePro'] < 11),
    df['YearsCodePro'] >= 11,
]
choices = ["Junior", "Mid-Level", "Senior", "Executive/Expert"]

df['ExperienceLevel'] = np.select(conditions, choices, default="Unknown")
df
```

### Task 1: Visualizing Data Composition with Pie Charts


##### 1.1 Create a Pie Chart of the Top 5 Databases Respondents Want to Work With


In the survey data, the `DatabaseWantToWorkWith` column lists the databases that respondents wish to work with. Let’s visualize the top 5 most-desired databases in a pie chart.



```python
##Write your code here
df1 = df[['DatabaseWantToWorkWith']]
df1['Database'] = df1['DatabaseWantToWorkWith'].str.split(';')
df1 = df1.explode('Database')
df1['Database'] = df1['Database'].str.strip()
df1_count = df1['Database'].value_counts().reset_index()
df1_count.columns = ['Database', 'Count']
top5 = df1_count.sort_values(by='Count', ascending=False).head(5)
top5.set_index('Database')['Count'].plot.pie(autopct='%1.1f%%', startangle=90)
plt.title('Distribution of top databases to want')
plt.ylabel('')
plt.show()
```

The `DevType` column lists the developer types for respondents. We’ll examine the distribution by showing the top 5 developer roles in a pie chart.



```python
##Write your code here
df2 = df[['DevType']]
df2['Developer'] = df2['DevType'].str.split(';')
df2 = df2.explode('Developer')
df2['Developer'] = df2['Developer'].str.strip()
df2_count = df2['Developer'].value_counts().reset_index()
df2_count.columns = ['Developer', 'Count']
top5_dev = df2_count.sort_values(by='Count', ascending=False).head(5)
top5_dev.set_index('Developer')['Count'].plot.pie(autopct='%1.1f%%', startangle=90)
plt.title('Distribution of top developer types')
plt.ylabel('')
plt.show()
```

##### 1.3 Create a pie chart for the operating systems used by respondents for professional use


The `OpSysProfessional` use column shows the operating systems developers use professionally. Let’s visualize the distribution of the top operating systems in a pie chart.



```python
##Write your code here
df3 = df[['OpSysProfessional use']]
df3['OpSys'] = df3['OpSysProfessional use'].str.split(';')
df3 = df3.explode('OpSys')
df3['OpSys'] = df3['OpSys'].str.strip()
df3_count = df3['OpSys'].value_counts().reset_index()
df3_count.columns = ['OpSys', 'Count']
top5_op = df3_count.sort_values(by='Count', ascending=False).head(5)
top5_op.set_index('OpSys')['Count'].plot.pie(autopct='%1.1f%%', startangle=90)
plt.title('Distribution of Top 5 Operating Systems')
plt.ylabel('')
plt.show()
```
